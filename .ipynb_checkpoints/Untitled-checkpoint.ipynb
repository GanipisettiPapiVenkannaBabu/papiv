```python
i=5
while i>=0:
    print(i)
    i-=1
```

```python
i=5
while i>=0:
    if i==1:
        break
    else:
        print(i)
        i-=1
else:
    print("completed")
```

```python
i=5
while i>=0:
    if i==1:
        continue
        i-=1
    else:
        print(i)
        i-=1
else:
    print("completed")
```
