```python
toy=input("types of toys must be in [battery, wire, wireless]:")
amount=int(input("enter the amount"))
if toy=="wireless":
    if amount>=2000:
        print("amount",amount-(amount*5 / 100))
    else :
        print("no discount")
elif toy == "battery":
     if amount>=500:
         print("amount",amount-(amount*10 / 100))
     else:
         print("no toys are free")
elif toy == "wire": 
    if amount>=1000:
        print("2toy free")
    elif amount >=500:
        print("extra toy ")
    else:
        print("not give the extra toy ")
else:
    print("invalid")
    
```

```python
n=int(input("enter a value:"))
for i in range(1,20):
    if i!=17:
        print(n,"*",i,"=",n*i)
    else:
        print("breaked")
        break
```

```python
n=int(input("enter a value:"))
for i in range(1,20):
    if i==17:
        break
    else:
        print(n,"*",i,"=",n*i)
     
```

```python
n=int(input("enter a value:"))
for i in range(1,11):
    if i%2==0:
        continue
    elif i>=6:
        break
    else:
        print(n,"*",i,"=",n*i)
```
