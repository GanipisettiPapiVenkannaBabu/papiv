```python
distance=int(input(" enter the distance"))
if distance<=0:
    print("invalid") 
elif distance<=5:
    price=distance*5
    print("price:",price)
elif distance<=12:
    price=5*5+(distance-5)*6
    print("price:",price)
elif distance<=20:
    price=(5*5)+(7*6)+(distance-12)*8
    print("price:",price)
elif distance<=30:
    price=(5*5)+(9*5)+(8*8)+(distance-20)*10
    print("price:",price)
else:
    print("ride not possible")
```
